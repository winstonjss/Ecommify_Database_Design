```python
from sqlalchemy import create_engine, text
from google.colab import userdata

def connect_to_supabase_postgres():
    print("Attempting to connect to Supabase PostgreSQL database...")
    try:
        # Retrieve Supabase PostgreSQL credentials from Colab secrets
        db_user = userdata.get('SUPABASE_DB_USER')
        db_password = userdata.get('SUPABASE_DB_PASSWORD')
        db_host = userdata.get('SUPABASE_DB_HOST') # e.g., 'aws-0-us-west-2.pooler.supabase.com'
        db_port = userdata.get('SUPABASE_DB_PORT') # e.g., '5432'
        db_name = userdata.get('SUPABASE_DB_NAME') # e.g., 'postgres'

        if not all([db_user, db_password, db_host, db_port, db_name]):
            print("Error: Supabase PostgreSQL credentials (user, password, host, port, name) are not set in Colab secrets. Please set them using the 🔑 Secrets tab.")
            return None

        # Construct the connection string
        # For Supabase, the format is typically postgresql://[user]:[password]@[host]:[port]/[database_name]
        connection_string = f"postgresql://{db_user}:{db_password}@{db_host}:{db_port}/{db_name}"

        # Create the engine
        engine = create_engine(connection_string)

        # Test the connection
        with engine.connect() as connection:
            result = connection.execute(text("SELECT 1"))
            print("Successfully connected to Supabase PostgreSQL database!")
            return engine

    except Exception as e:
        print(f"Could not connect to Supabase PostgreSQL database: {e}")
        return None

if __name__ == "__main__":
    # Example usage:
    engine = connect_to_supabase_postgres()
    if engine:
        print("Database connection successful. You can now use the 'engine' object for queries.")
        # Example: Fetching data
        # with engine.connect() as connection:
        #     result = connection.execute(text("SELECT * FROM your_table_name LIMIT 5"))
        #     for row in result:
        #         print(row)
        # Close the engine when done
        #csv_to_insert(engine, "olist_products_dataset")
        insert_geolocation_data(engine, table_name='geolocations', csv_file_path='./e-commerce/olist_geolocation_dataset.csv', chunk_size=5000)

        engine.dispose()
    else:
        print("Failed to connect to the database.")
```
